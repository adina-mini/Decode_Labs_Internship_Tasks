```python
# PROJECT 3: AI Recommendation System - Tech Stack Recommender
# DecodeLabs | Adina Rehman

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
job_roles = {
    "Data Scientist":        "python machine learning sql data analysis statistics numpy pandas",
    "AI Engineer":           "python deep learning tensorflow pytorch neural networks automation",
    "Backend Developer":     "python java sql apis rest databases nodejs docker",
    "Frontend Developer":    "javascript react css html tailwind ui ux design",
    "DevOps Engineer":       "docker kubernetes aws cloud ci cd linux automation git",
    "ML Engineer":           "python machine learning mlops docker kubernetes tensorflow pipelines",
    "Data Analyst":          "sql excel python data visualization power bi tableau statistics",
    "Cloud Architect":       "aws azure cloud docker kubernetes infrastructure automation",
    "Cybersecurity Analyst": "networking linux security ethical hacking python firewalls",
    "Full Stack Developer":  "javascript react nodejs python sql html css apis",
}
```

```python
print("=== TECH STACK RECOMMENDER ===")
print("Enter your top 3 skills (e.g: python, machine learning, automation)\n")

skill1 = input("Skill 1: ").strip().lower()
skill2 = input("Skill 2: ").strip().lower()
skill3 = input("Skill 3: ").strip().lower()
user_profile = f"{skill1} {skill2} {skill3}"
```

```python
role_names = list(job_roles.keys())
role_descriptions = list(job_roles.values())
```

```python
# combine user profile with all roles into one corpus
corpus = role_descriptions + [user_profile]
```

```python

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)
```

```python

# user vector is the last one
user_vector = tfidf_matrix[-1]
role_vectors = tfidf_matrix[:-1]
scores = cosine_similarity(user_vector, role_vectors)[0]
```

```python
ranked = sorted(zip(role_names, scores), key=lambda x: x[1], reverse=True)
top3 = ranked[:3]
```

```python
# ── OUTPUT ──
print(f"\n Skills entered: {skill1}, {skill2}, {skill3}")
print("\n TOP 3 RECOMMENDED CAREER PATHS:\n")
for i, (role, score) in enumerate(top3, 1):
    print(f"  {i}. {role}  —  Match Score: {score*100:.1f}%")
```
