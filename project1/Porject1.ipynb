```python
# PROJECT 1: Rule-Based AI Chatbot - UPGRADED
# DecodeLabs Industrial Training | Adina Rehman

responses = {
    ("hello", "hi", "hey", "sup", "assalam"): "Hey! I'm DecodeBot 🤖 How can I help?",
    ("how are you", "how r you", "hows it going", "you good"): "Running at 100% efficiency! What do you need?",
    ("what is ai", "explain ai", "define ai", "tell me about ai"): "AI = machines simulating human intelligence. The future is now.",
    ("what is ml", "what is machine learning", "explain ml"): "ML is AI that learns from data instead of hardcoded rules.",
    ("what is decodelabs", "tell me about decodelabs", "decode labs"): "DecodeLabs is your industrial AI training ground. ",
    ("who made you", "who built you", "your creator"): "Adina Rehman — built as DecodeLabs Project 1. 🔥",
    ("bye", "goodbye", "later", "cya"): "Goodbye! Keep building. 🚀",
    ("help", "what can you do", "commands"): "Ask me about AI, ML, DecodeLabs, or just vibe.",
    ("thanks", "thank you", "thx"): "Anytime! That's what I'm here for.",
}

def get_response(user_input):
    for keywords, reply in responses.items():
        for keyword in keywords:
            if keyword in user_input:  # substring match — not exact!
                return reply
    return "Hmm, I don't know that yet. Try asking about AI or ML!"

print("DecodeBot initialized. Type 'exit' to terminate.\n")

while True:
    raw = input("You: ")
    clean = raw.lower().strip()

    if clean in ("exit", "quit", "q"):
        print("Bot: Session terminated. Goodbye! 👋")
        break

    print(f"Bot: {get_response(clean)}\n")
```
